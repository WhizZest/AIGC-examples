```python
from transformers import BarkModel, AutoProcessor
import torch
import scipy

# 检查设备
device = "cuda" if torch.cuda.is_available() else "cpu"

# 加载模型和处理器
model = BarkModel.from_pretrained("suno/bark").to(device)
processor = AutoProcessor.from_pretrained("suno/bark")

# 处理文本
text_prompt = "Hello, this is Bark speaking! [laughs] How are you doing today?"
inputs = processor(text_prompt, voice_preset="v2/en_speaker_6", return_tensors="pt").to(device)

# 生成音频
with torch.no_grad():
    audio_array = model.generate(**inputs, do_sample=True)
    
# 转换为 numpy 数组
audio_array = audio_array.cpu().numpy().squeeze()

# 保存为 WAV 文件
sampling_rate = model.generation_config.sample_rate
scipy.io.wavfile.write("bark_output.wav", rate=sampling_rate, data=audio_array)

print(f"Audio saved as bark_output.wav with sampling rate {sampling_rate} Hz")
# 在jupyter notebook中播放音频
from IPython.display import Audio
Audio("bark_output.wav")
```

```python
from transformers import BarkModel, AutoProcessor
import torch
import scipy

device = "cuda" if torch.cuda.is_available() else "cpu"

# 加载模型
model = BarkModel.from_pretrained("suno/bark").to(device)
processor = AutoProcessor.from_pretrained("suno/bark")

# voice_preset 在 processor 中指定
text = "你好，世界！"
inputs = processor(
    text, 
    voice_preset="v2/zh_speaker_1",  # 这里指定voice_preset
    return_tensors="pt"
).to(device)

# 生成音频（不需要传递voice_preset）
audio_array = model.generate(**inputs)

# 保存
sampling_rate = model.generation_config.sample_rate
scipy.io.wavfile.write("output.wav", rate=sampling_rate, data=audio_array.cpu().numpy().squeeze())
# 在jupyter notebook中播放音频
from IPython.display import Audio
Audio("output.wav", autoplay=True)
```
