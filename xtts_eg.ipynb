```python
from TTS.api import TTS
import torch
import numpy as np

# 加载支持GST的模型
tts = TTS("tts_models/zh-CN/baker/tacotron2-DDC-GST")

# 1. 基本控制：通过文本内容暗示风格
styles_by_text = {
    "中性": "系统正常运行，一切参数都在标准范围内。",
    "高兴": "太棒了！今天真是个令人兴奋的好日子！",
    "严肃": "请注意，这是一条重要通知，请务必仔细听。",
    "温和": "别担心，慢慢来，我会一直在这里帮助你。"
}

# 2. 高级控制：直接设置GST向量（细粒度控制）
def generate_with_style(text, style_vector=None, speaker_id=None):
    # 获取模型内部对象以访问低级API
    model = tts.synthesizer.tts_model
    model.eval()
    
    # 准备输入
    inputs = tts.synthesizer.tokenizer.text_to_ids(text, language="zh-cn")
    
    # 设置风格向量（GST）
    if style_vector is not None:
        if isinstance(style_vector, str):
            # 预设风格向量
            presets = {
                "happy": torch.randn(1, 1, 128) * 0.3 + 0.5,  # 高兴
                "serious": torch.randn(1, 1, 128) * 0.1 - 0.2,  # 严肃
                "neutral": torch.zeros(1, 1, 128),  # 中性
                "female": torch.randn(1, 1, 128) * 0.2 + 0.1,  # 偏女性
                "male": torch.randn(1, 1, 128) * 0.2 - 0.1,  # 偏男性
            }
            style_vector = presets.get(style_vector, presets["neutral"])
    
    # 合成语音
    with torch.no_grad():
        output = model.inference(inputs, style_vector=style_vector)
    
    return output["wav"]

# 生成不同风格的语音
for style_name, text in styles_by_text.items():
    output_file = f"reference_{style_name}.wav"
    tts.tts_to_file(text=text, file_path=output_file)
    print(f"生成 {style_name} 风格: {output_file}")
```

```python
from TTS.api import TTS
import hashlib
import numpy as np

tts = TTS("tts_models/zh-CN/baker/tacotron2-DDC-GST")

# 多次生成
audio1 = tts.tts("测试文本")
audio2 = tts.tts("测试文本")
audio3 = tts.tts("测试文本")

# 确保是numpy数组
if isinstance(audio1, list):
    audio1 = np.array(audio1, dtype=np.float32)
    audio2 = np.array(audio2, dtype=np.float32)
    audio3 = np.array(audio3, dtype=np.float32)

# 方法1：转换为bytes再计算哈希（正确方式）
hash1 = hashlib.md5(audio1.tobytes()).hexdigest()
hash2 = hashlib.md5(audio2.tobytes()).hexdigest()
hash3 = hashlib.md5(audio3.tobytes()).hexdigest()

print("=== 传统TTS确定性测试 ===")
print(f"音频1哈希: {hash1}")
print(f"音频2哈希: {hash2}")
print(f"音频3哈希: {hash3}")

if hash1 == hash2 == hash3:
    print("✅ 完全相同 - 传统TTS是确定性的")
else:
    print("❌ 不同 - 检查模型设置")
```
